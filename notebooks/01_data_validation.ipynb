```python
import pandas as pd
from pathlib import Path

DATA = Path("../data/raw")

patients = pd.read_csv(DATA / "patients.csv")
visits = pd.read_csv(DATA / "dental_visits.csv")
notes = pd.read_csv(DATA / "clinical_notes.csv")
referrals = pd.read_csv(DATA / "referrals.csv")
pathology = pd.read_csv(DATA / "pathology.csv")
appointments = pd.read_csv(DATA / "appointments.csv")
outcomes = pd.read_csv(DATA / "outcomes.csv")

print(f"patients      : {len(patients):>6,}")
print(f"visits        : {len(visits):>6,}")
print(f"notes         : {len(notes):>6,}")
print(f"referrals     : {len(referrals):>6,}")
print(f"pathology     : {len(pathology):>6,}")
print(f"appointments  : {len(appointments):>6,}")
print(f"outcomes      : {len(outcomes):>6,}")
```

```python
# Sanity check: any unexpected NaNs?
print("NaN counts in patients:")
print(patients.isna().sum())
```

```python
# Cancer prevalence by risk factor — does the signal look right?
print("Overall cancer prevalence:", f"{patients['oral_cancer_positive'].mean()*100:.1f}%")
print()
print("By smoking status:")
print(patients.groupby("smoking_status")["oral_cancer_positive"].mean().round(3))
print()
print("By alcohol use:")
print(patients.groupby("alcohol_use")["oral_cancer_positive"].mean().round(3))
print()
print("By HPV status:")
print(patients.groupby("hpv_status")["oral_cancer_positive"].mean().round(3))
```
